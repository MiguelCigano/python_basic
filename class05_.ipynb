```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.io.wavfile as waves
```

Senal de audio

```python
audio_file = 'Alarm01.wav'
```

Procesamiento

```python
fs, sound = waves.read(audio_file)
```

```python
print("fs: ", fs)
print("dim sound: ", np.shape(sound))
print(sound)
```

```python
#segmento de tiempo

chanel  = 0
init    = 2600
finish  = 2800
```

```python
#extraccion de datos
dt = 1/fs
t = np.arange(init*dt, finish*dt, dt)
samples = len(t)
segment = sound[init:init + samples, chanel]
```

```python
#out
plt.plot(t, segment)
plt.grid(True)
plt.show()
```
