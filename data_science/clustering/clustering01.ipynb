```python
# with work with distance
from scipy.spatial import distance_matrix
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import plotly.express as px
import plotly.graph_objects as go
import numpy as np
```

```python
df = pd.read_csv('../datasets/movies/movies.csv', sep = ';')
```

```python
movies = df.columns.values.tolist()[1:]
movies
```

```python
# Mahathan
dd1 = distance_matrix(df[movies], df[movies], p =  1)
# Euclidian
dd2 = distance_matrix(df[movies], df[movies], p =  2)
# Minkowski
dd10 = distance_matrix(df[movies], df[movies], p = 10)
```

```python
# Normal dd is a array, need to change into df
def dm_to_df(dd, col_name):
    # import pandas as pd, por que hacer esto?
    return pd.DataFrame(dd, index = col_name, columns = col_name)
```

```python
dm_to_df(dd1, df['user_id'])
```

```python
dm_to_df(dd2, df['user_id'])
```

```python
dm_to_df(dd10, df['user_id'])
```

```python
user_ids = df['user_id'].tolist()

# 2. Crear la superficie
fg = go.Figure(data=[go.Surface(
    z=dd10, 
    x=user_ids, 
    y=user_ids,
    colorscale='magma' # El equivalente exacto a :magma de Julia
)])

# 2. Configuración estética (Títulos y etiquetas)
fg.update_layout(
    title='Mapa de Distancias de Minkowski (p=10)',
    scene=dict(
        xaxis_title='Usuario X',
        yaxis_title='Usuario Y',
        zaxis_title='Distancia'
    ),
    autosize=False,
    width=800,
    height=800,
    margin=dict(l=65, r=50, b=65, t=90)
)

fg.show()
```
