```python
import numpy as np
import math as mth
import matplotlib.pyplot as plt
```

```python
# parámetros
f = 2                  # frecuencia en Hz
omega = 2 * np.pi * f  # frecuencia angular

# tiempo
t = np.linspace(0, 2, 1000)

# señales
y_f = np.sin(2 * np.pi * f * t)
y_omega = np.sin(omega * t)

# gráfica
plt.figure(figsize=(10, 5))
plt.plot(t, y_f, label=r"$\sin(2\pi f t)$", linestyle='--')
plt.plot(t, y_omega, label=r"$\sin(\omega t)$", alpha=0.5)
plt.title("Comparación entre $\omega = 2\pi f$")
plt.xlabel("Tiempo (s)")
plt.ylabel("Amplitud")
plt.grid(True)
plt.legend()
```

```python
# señal continua
t = np.arange(0, 6 * np.pi, 0.1)
amplitud = np.sin(t)

# gráfica
plt.figure(figsize=(10, 5))
plt.plot(t, amplitud, label = "Señal continua")

plt.grid(True)
plt.xlabel("Tiempo (t)")
plt.ylabel("sin(t)")
plt.legend()
```

```python
# t2 = np.linspace(0, 6 * np.pi, 100)
t2 = np.linspace(0, 30, 100)
amplitud2 = np.exp(-0.1*t2) * np.sin(t2)

# gráfica 
plt.figure(figsize=(10, 5))
plt.plot(t2, amplitud2, label = "Señal continua")
plt.grid(True)
plt.xlabel("Tiempo (s)")
plt.ylabel("Amplitud")
plt.legend()
```

```python
# tiempo
t3 = np.arange(0,4*np.pi, 0.1)

y1 = np.sin(2*np.pi*t3)
y2 = np.cos(0.5*np.pi*t3)
y3 = mth.e**(-0.4*t3) * (y1 + y2)

# gráfica 
plt.figure(figsize=(10, 5))
plt.plot(t3, y3)
plt.plot(t3, y2)
plt.plot(t3, y1)

plt.grid(True)
plt.xlabel("t")
plt.ylabel("y(t)")
plt.axhline(0, color = "black")
plt.axvline(0, color="blue")
```

```python

# parámetros
f0 = 2  # frecuencia fundamental (Hz)
t = np.linspace(0, 2, 1000)  # tiempo en segundos

# componentes: fundamental y armónicos
y1 = np.sin(2 * np.pi * f0 * t)          # fundamental
y2 = 0.5 * np.sin(2 * np.pi * 2 * f0 * t)  # 2do armónico
y3 = 0.3 * np.sin(2 * np.pi * 3 * f0 * t)  # 3er armónico

# señal compuesta
y_total = y1 + y2 + y3

# gráfico
plt.figure(figsize=(12, 6))

# señal compuesta
plt.plot(t, y_total, label="Señal compuesta", color="black", linewidth=2)

# componentes individuales
plt.plot(t, y1, "--", label="Fundamental (2 Hz)")
plt.plot(t, y2, "--", label="2° armónico (4 Hz)")
plt.plot(t, y3, "--", label="3° armónico (6 Hz)")

plt.title("Construcción de una señal con frecuencia fundamental y armónicos")
plt.xlabel("Tiempo (s)")
plt.ylabel("Amplitud")
plt.grid(True)
plt.legend()
plt.tight_layout()
```

```python
# tiempo de 0 a 10 segundos
t = np.linspace(0, 10, 1000)
x1 = np.sin(2 * np.pi * t)
x2 = np.sin(2 * np.pi * np.sqrt(2) * t)
x = x1 + x2

plt.figure(figsize=(10, 4))
plt.plot(t, x, label='x(t) = sin(2πt) + sin(2π√2 t)')
plt.title("Suma de señales con frecuencias no conmensurables")
plt.xlabel("Tiempo (s)")
plt.ylabel("Amplitud")
plt.grid(True)
plt.legend()
```

## Teorema de Nyquist-Shannon

```python
# señal continua (densamente muestreada)
t4 = np.arange(0, 4*np.pi, 0.001)
y1 = np.sin(2*np.pi*t4)               # frecuencia f1 = 1 Hz
y2 = np.cos(0.5*np.pi*t4)             # frecuencia f2 = 0.25 Hz
y3 = np.exp(-0.4*t4) * (y1 + y2)      # señal compuesta atenuada
```

```python
plt.figure(figsize=(10, 5))
plt.plot(t4, y3, label="Señal compuesta atenuada")
plt.grid(True)
plt.xlabel("Tiempo (s)")
plt.ylabel("y(t)")
plt.legend()
```

```python
fs_good = 4  # Hz
t_good = np.arange(0, 4*np.pi, 1/fs_good)
y_good = np.exp(-0.4*t_good) * (np.sin(2*np.pi*t_good) + np.cos(0.5*np.pi*t_good))
```

```python
fs_bad = 1  # Hz < 2 Hz → violamos Nyquist
t_bad = np.arange(0, 4*np.pi, 1/fs_bad)
y_bad = np.exp(-0.4*t_bad) * (np.sin(2*np.pi*t_bad) + np.cos(0.5*np.pi*t_bad))
```

```python
# tamaño de gráfica
plt.figure(figsize=(12, 5))

# señal continua
plt.plot(t4, y3, 'k-', label="Original (continua)")

# muestreo malo
plt.stem(t_bad, y_bad, linefmt='r--', markerfmt='ro', basefmt=" ", label="Muestreo malo (fs = 1 Hz)")

plt.grid(True)
plt.xlabel("t")
plt.ylabel("y(t)")
plt.axhline(0, color="black", linewidth=0.5)
plt.title("aliasing")
plt.legend()
```

```python
# tamaño de gráfica
plt.figure(figsize=(12, 5))

# señal continua
plt.plot(t4, y3, 'k-', label="Original (continua)")

# muestreo bueno
plt.stem(t_good, y_good, linefmt='g-', markerfmt='go', basefmt=" ", label="Muestreo bueno (fs = 10 Hz)")

plt.grid(True)
plt.xlabel("t")
plt.ylabel("y(t)")
plt.axhline(0, color="black", linewidth=0.5)
plt.title("Comparación de muestreo: bueno vs. aliasing")
plt.legend()
```

```python
plt.figure(figsize=(12, 5))

# señal continua
plt.plot(t4, y3, 'k-', label="Original (continua)")

# muestreo bueno
plt.stem(t_good, y_good, linefmt='g-', markerfmt='go', basefmt=" ", label="Muestreo bueno (fs = 10 Hz)")

# muestreo malo
plt.stem(t_bad, y_bad, linefmt='r--', markerfmt='ro', basefmt=" ", label="Muestreo malo (fs = 1 Hz)")

plt.grid(True)
plt.xlabel("t")
plt.ylabel("y(t)")
plt.axhline(0, color="red", linewidth=0.5)
plt.title("Comparación de muestreo: bueno vs. aliasing")
plt.legend()
```
