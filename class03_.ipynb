```python
import numpy as np
import matplotlib.pyplot as plt

# parámetros de la señal original
fs = 100  # frecuencia de muestreo alta (Hz) para evitar aliasing
T = 2     # duración de la señal (segundos)
t = np.arange(0, T, 1/fs)

# señal: suma de dos frecuencias
x = np.sin(6*np.pi*t) + 0.5*np.cos(2*np.pi*t)

# FFT
X = np.fft.fft(x)
freqs = np.fft.fftfreq(len(t), 1/fs)
idx = freqs >= 0

# Gráfico del espectro
plt.figure(figsize=(12, 4))
plt.stem(freqs[idx], np.abs(X[idx]) / len(x), use_line_collection=True)
plt.title("Espectro en frecuencia sin aliasing (fs = 100 Hz)")
plt.xlabel("Frecuencia (Hz)")
plt.ylabel("Amplitud")
plt.grid(True)
plt.tight_layout()


#-----------------------------------
```

```python
# Ahora con aliasing: fs bajo
fs_alias = 4  # Menor que 2*f_max
T_alias = 2

# Nueva señal muestreada con menos frecuencia
t_alias = np.arange(0, T_alias, 1/fs_alias)
x_alias = np.sin(6*np.pi*t_alias) + 0.5*np.cos(2*np.pi*t_alias)

# FFT con aliasing
X_alias = np.fft.fft(x_alias)
freqs_alias = np.fft.fftfreq(len(t_alias), 1/fs_alias)
idx_alias = freqs_alias >= 0

# Gráfico con aliasing
plt.figure(figsize=(12, 4))
plt.stem(freqs_alias[idx_alias], np.abs(X_alias[idx_alias]) / len(x_alias), use_line_collection=True)
plt.title("Espectro en frecuencia con aliasing (fs = 4 Hz)")
plt.xlabel("Frecuencia (Hz)")
plt.ylabel("Amplitud")
plt.grid(True)
plt.tight_layout()
plt.show()
```

Discretizando una senal de audio


```python
fs = 100  # Alta frecuencia de muestreo
t = np.arange(0, 2, 1/fs)
x = np.sin(6*np.pi*t) + 0.5*np.cos(2*np.pi*t)


X = np.fft.fft(x)
freqs = np.fft.fftfreq(len(t), 1/fs)
idx = freqs >= 0

plt.plot(freqs[idx], np.abs(X[idx]) / len(x))
plt.title("Espectro en frecuencia sin aliasing (fs = 100 Hz)")
plt.xlabel("Frecuencia (Hz)")
plt.ylabel("Amplitud")
plt.grid(True)
plt.show()
```
